```python
import cfgrib

file = "/Users/ljob/Desktop/flxf06.gdas.200311.grb2"

ds = cfgrib.open_dataset(
    file,
    engine="cfgrib",
    filter_by_keys={
                "typeOfLevel": "heightAboveGround",
                "level": 2,
            },
    decode_timedelta=False,
)
print(ds)
print("Variables:")
for var in ds.data_vars:
    print(f"  {var}")

print("\nDimensions:")
print(ds.dims)

print("\nCoordinates:")
for coord in ds.coords:
    print(f"  {coord}")
```
