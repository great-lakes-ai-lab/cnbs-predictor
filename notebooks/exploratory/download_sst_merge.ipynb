```python
import pandas as pd

# Read file
df = pd.read_csv("/Users/ljob/Downloads/all_year_glsea_avg_s_C.csv")

# Rename first column to day_of_year
df = df.rename(columns={df.columns[0]: "day_of_year"})

# Convert from wide → long format
df_long = df.melt(
    id_vars="day_of_year",
    var_name="year",
    value_name="superior_sst"
)

# Create proper datetime column
df_long["date"] = pd.to_datetime(
    df_long["year"].astype(str)
    + df_long["day_of_year"].astype(int).astype(str).str.zfill(3),
    format="%Y%j"
)

# Keep only desired columns
df_superior = df_long[["date", "superior_sst"]]

# Sort by date
df_superior = df_superior.sort_values("date").reset_index(drop=True)

# Convert from Celsius to Kelvin
df_superior['superior_sst'] = df_superior['superior_sst'] + 273.15

print(df_superior)
```

```python
import pandas as pd

# Read file
df = pd.read_csv("/Users/ljob/Downloads/all_year_glsea_avg_o_C.csv")

# Rename first column to day_of_year
df = df.rename(columns={df.columns[0]: "day_of_year"})

# Convert from wide → long format
df_long = df.melt(
    id_vars="day_of_year",
    var_name="year",
    value_name="ontario_sst"
)

# Create proper datetime column
df_long["date"] = pd.to_datetime(
    df_long["year"].astype(str)
    + df_long["day_of_year"].astype(int).astype(str).str.zfill(3),
    format="%Y%j"
)

# Keep only desired columns
df_long = df_long[["date", "ontario_sst"]]

# Sort by date
df_ontario = df_long.sort_values("date").reset_index(drop=True)

# Convert from Celsius to Kelvin
df_ontario['ontario_sst'] = df_ontario['ontario_sst'] + 273.15

print(df_ontario)
```

```python
# Merge all lake SST datasets on date
df_sst = (
    df_superior
    .merge(df_michigan, on="date", how="outer")
    .merge(df_huron, on="date", how="outer")
    .merge(df_erie, on="date", how="outer")
    .merge(df_ontario, on="date", how="outer")
)

# Calculate combined Michigan-Huron mean SST
df_sst["michigan-huron_sst"] = (
    df_sst[["michigan_sst", "huron_sst"]]
    .mean(axis=1)
)

# Drop individual Michigan and Huron columns
df_sst = df_sst.drop(columns=["michigan_sst", "huron_sst"])

# Optional: reorder columns
df_sst = df_sst[
    [
        "date",
        "superior_sst",
        "michigan-huron_sst",
        "erie_sst",
        "ontario_sst",
    ]
]

print(df_sst)
```

```python
df_sst = df_sst.dropna()
```
